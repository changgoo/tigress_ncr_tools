```python
from pathena.pdf2d_reader import plot_all, read_all_pdfs, plot_column_density
```

```python
basedir = "/anvil/scratch/x-ckim5/TIGRESS-NCR/R8_8pc_NCR_row0000"
problem_id = "R8_8pc_NCR"
```

```python
pdf_dict = read_all_pdfs(basedir, problem_id)
```

```python
import matplotlib.pyplot as plt
for num in range(0, len(pdf_dict["x1-x2"]), 10):
    pdf = pdf_dict["x1-x2"][num]
    print(f"Frame {num}: {(pdf['weights']['nH']/64).mean()}")
    plt.scatter(num, (pdf['weights']['nH']/pdf['weights']['volume']).mean(), color='k', s=10)
```

```python
for pdf_id, pdf in pdf_dict.items():
    projection = (pdf_id == "x1-x2") or (pdf_id == "x1-x3")
    fig = plot_all(pdf[1], projection=projection)
```
